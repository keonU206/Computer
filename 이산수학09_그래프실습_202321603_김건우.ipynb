## 그래프 탐색 실습(DFS, BFS, 친구 추천)

- 그래프를 인접 리스트로 표현할 수 있다.
  - 다만, 파이썬에서는 이와 유사한 딕셔너리 자료구조를 제공한다. 구현의 간편함 측면에서 딕셔너리를 사용한다.
- DFS와 BFS의 동작 원리를 이해하고 구현할 수 있다.
- BFS를 활용한 실생활 응용(친구 추천)의 아이디어를 익힐 수 있다.

```python
# 예시 그래프: 친구 관계
graph = {
    'Alice': ['Bob', 'Charlie'],
    'Bob': ['David', 'Eve'],
    'Charlie': ['Frank'],
    'David': [],
    'Eve': ['Frank'],
    'Frank': []
}
```

## DFS 구현


```python
def dfs(graph, v, visited):
    visited.add(v)
    print(v, end=' ')
    for neighbor in graph[v]:
        if neighbor not in visited:
            dfs(graph, neighbor, visited)
```

```python
print("DFS 탐색 순서:")
dfs(graph, 'Alice', set())
```

## 2. BFS 탐색 구현


```python
# 큐(Queue)를 효율적으로 구현할 수 있는 도구
from collections import deque

def bfs(graph, start):
    visited = set()
    queue = deque([start])
    visited.add(start)

    while queue:
        v = queue.popleft()
        print(v, end=' ')
        for neighbor in graph[v]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)

print("\nBFS 탐색 순서:")
bfs(graph, 'Alice')
```

## 3. 친구 추천 시스템(BFS 응용)

- 2단계 거리 친구만 추천(내 친구의 친구)
- 나와 직접 연결된 친구는 제외

```python
from collections import deque

def recommend_friend(graph, start):
    visited = set()
    queue = deque([(start, 0)])  # (노드, 거리)

    print(f"\n{start}의 친구 추천 (2단계):")

    while queue:
        current, depth = queue.popleft()
        if current not in visited:
            visited.add(current)
            if depth == 2:
                print("-", current)
            if depth < 2:
                for neighbor in graph[current]:
                    queue.append((neighbor, depth + 1))
```

```python
recommend_friend(graph, 'Alice')
```
